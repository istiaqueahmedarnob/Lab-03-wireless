```python
import numpy as np
import matplotlib.pyplot as plt

C = 420
R = 1.0
N = np.array([3, 4, 7, 9, 12])
K_INTERFERERS = 6

reuse_ratio = np.sqrt(3 * N)
D = R * reuse_ratio
channels_per_cell = C / N

n_pl = 4
sir_db = 10 * np.log10((reuse_ratio ** n_pl) / K_INTERFERERS)

print("=" * 62)
print(f"{'N':>3} | {'D/R':>6} | {'D (km)':>7} | {'Channels/cell':>13} | {'SIR (dB)':>8}")
print("-" * 62)
for n, ratio, dist, ch, s in zip(N, reuse_ratio, D, channels_per_cell, sir_db):
    print(f"{n:>3} | {ratio:>6.2f} | {dist:>7.2f} | {ch:>13.1f} | {s:>8.2f}")
print("=" * 62)

plt.figure()
plt.plot(N, D, "o-", color="tab:blue")
for n, d in zip(N, D):
    plt.annotate(f"{d:.2f}", (n, d), textcoords="offset points", xytext=(0, 8), ha="center")
plt.xlabel("Cluster Size, N")
plt.ylabel("Reuse Distance D (km)")
plt.title("Reuse Distance vs Cluster Size")
plt.grid(True)
plt.savefig("reuse_distance_plot.png", dpi=150, bbox_inches="tight")

plt.figure()
plt.plot(N, channels_per_cell, "s-", color="tab:red")
for n, ch in zip(N, channels_per_cell):
    plt.annotate(f"{ch:.0f}", (n, ch), textcoords="offset points", xytext=(0, 8), ha="center")
plt.xlabel("Cluster Size, N")
plt.ylabel("Channels per Cell")
plt.title("Channels per Cell vs Cluster Size")
plt.grid(True)
plt.savefig("channels_per_cell_plot.png", dpi=150, bbox_inches="tight")

fig, ax1 = plt.subplots()
ax1.plot(N, D, "o-", color="tab:blue")
ax1.set_xlabel("Cluster Size, N")
ax1.set_ylabel("Reuse Distance D (km)", color="tab:blue")
ax2 = ax1.twinx()
ax2.plot(N, channels_per_cell, "s--", color="tab:red")
ax2.set_ylabel("Channels per Cell", color="tab:red")
ax1.grid(True)
plt.title("Capacity-Interference Trade-off")
plt.savefig("tradeoff_plot.png", dpi=150, bbox_inches="tight")


def smallest_feasible_N(N_candidates, R, C, D_min):
    for n in np.sort(N_candidates):
        d = R * np.sqrt(3 * n)
        if d >= D_min:
            return int(n), d, C / n
    return None


D_MIN = 4.0
result = smallest_feasible_N(N, R, C, D_MIN)

print(f"\nDesign challenge: C = {C}, R = {R} km, D_min = {D_MIN} km")
if result is None:
    print("No candidate N satisfies the reuse-distance requirement.")
else:
    n_sel, d_sel, ch_sel = result
    print(f"Smallest feasible N = {n_sel}")
    print(f"  D/R = {np.sqrt(3 * n_sel):.2f}, D = {d_sel:.2f} km, "
          f"channels/cell = {ch_sel:.1f}")

plt.show()
```
